```python
#simple graph with one node
import networkx as nx
import matplotlib.pyplot as plt
G = nx.Graph()

G.add_node("I")                          #add a new node
G.add_nodes_from(["E", "F","G","H"])     #multiple node at once in the form of a list
G.add_nodes_from([                       #nodes with attributes
    ("A",{"age":0, "color":"red"}),
    ("B",{"age":10, "color":None}),
    ("C",{"age":12, "color":"green"}),
    ("D",{"age":13, "color":"yellow"})
])
G.add_edge("A","B")                                   #adding an edge
G.add_edges_from([                                    #adding multiple edges with attributes
    ("A","E",{"weight":5, "color": "red"}),
    ("A","F", {"weight": 10, "relation":"friend"}),
    ("C","D"), ("B","C"), ("E","F"), ("D","E"),("E","B"), ("G","H")])

p = {"A": (2,2),                  #position of each node on a plane
     "B": (3,2.5),  "C":(4,3), 
     "D":(4,4),     "E":(3,4.5), 
     "F":(2,5),     "G":(3,5), 
     "H":(4,5),     "I":(5,5),
     } 

nx.draw(
    G,
    with_labels=True,
    pos=p,
    node_size = 1000,
    node_color='skyblue',
    edge_color="lightgrey",
    width = 3.0,
    font_size = 18,
    font_weight= "bold",
  )
plt.title("Simple Graph")
plt.show()
```

```python
#custom ndoe labels
my_node_labels = {
    "A": "Alice",
    "B": "Bob",
    "C": "Charlie",
    "D": "David",
    "E": "Eve",
    "F": "Frank",
    "G": "Grace",
    "H": "Hannah"
}
nx.draw(
    G,
    with_labels=True,
    labels=my_node_labels,
    pos=p,
    width = 3.0,
    edge_color="lightgrey",
    node_size=1500,
    font_size=12,
)
```

```python
G.nodes #returns a list of all the nodes in the graph
```

```python
G.nodes(data=True) #returns a list of all the nodes with their attributes in the form of a dictionary
```

```python
G.degree() #returns the degree of each node in the graph
```

```python
G.adj #returns the adjacency list of the graph in the form of a dictionary where the keys are the nodes and the values are the list of adjacent nodes
```

```python
G.edges #returns a list of all the edges in the graph
```

```python
G.edges.data("weight") #returns a list of all the edges with their weight attribute in the form of a tuple (node1, node2, weight)
```

```python
sm = 0
for u,v,w in G.edges.data("weight"):  #traverse through edges and print weights
  print(u,v,w)
  if w is None:
    continue
  sm = sm + w
print("Sum of weights is :",sm)
```
